Notebook for writing pulsar names to a file.

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
from pathlib import Path

data_path = Path('../datasets/ng20_v1p1_dmx_feathers')
feather_files = list(data_path.glob('*.feather'))

psr_names = []
for feather_file in feather_files:
    psr_name = feather_file.stem.split('_')[0].strip()  # remove \r, \n, spaces
    psr_names.append(psr_name)

# Force Unix line endings even on Windows
with open('../datasets/psr_names.txt', 'w', newline='\n', encoding='utf-8') as f:
    for psr_name in psr_names:
        f.write(psr_name + '\n')
```
